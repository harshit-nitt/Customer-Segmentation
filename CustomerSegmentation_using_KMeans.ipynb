```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
%matplotlib inline
```

```python
customer_data = pd.read_csv("/content/Mall_Customers.csv")
```

##choosing annual income and spending score column

```python
X=customer_data.iloc[:,[3,4]].values
```

```python
print(X)
```

```python
wcss = []

for i in range(1,11):
  kmeans = KMeans(n_clusters = i , init="k-means++", random_state=42)
  kmeans.fit(X)
  wcss.append(kmeans.inertia_)
```

###plotting elbow curve

```python
sns.set()
plt.plot( range(1,11), wcss)
plt.title("elbow curve")
plt.xlabel("no of clusters")
plt.ylabel("WCSS")
plt.show()
```

Optimum Number of clusters = 5

```python
#Training the KMeans Clustering Model

kmeans = KMeans(n_clusters = 5, init = 'k-means++')
#return a label for each data point on basis of it's cluster
Y= kmeans.fit_predict(X)
print(Y)
```

```python
#Visualising all clusters

plt.figure(figsize=(8,8))

plt.scatter(X[Y==0,0], X[Y==0,1], s = 50, c = 'green', label = "cluster-1" )
plt.scatter(X[Y==1,0], X[Y==1,1], s = 50, c = 'red', label = "cluster-1" )
plt.scatter(X[Y==2,0], X[Y==2,1], s = 50, c = 'blue', label = "cluster-1" )
plt.scatter(X[Y==3,0], X[Y==3,1], s = 50, c = 'yellow', label = "cluster-1" )
plt.scatter(X[Y==4,0], X[Y==4,1], s = 50, c = 'pink', label = "cluster-1" )


##plot the centroids
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], s =100, c = "cyan",label = 'centroids')

plt.title("Customer Group")
plt.xlabel("annual income")
plt.ylabel("spending score")
plt.show()
```
